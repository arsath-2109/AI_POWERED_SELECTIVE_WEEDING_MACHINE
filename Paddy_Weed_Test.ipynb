```python

!pip -q install gdown

import gdown
import zipfile
import os

file_id = "1pQ1rbvE2c9JOzFUAkb_HN8jgUq6QCt2Q"
zip_path = "/content/paddy-Weed.v1i.yolov8-obb.zip"
extract_path = "/content/paddy_weed_dataset"

url = f"https://drive.google.com/uc?id={file_id}"

gdown.download(url, zip_path, quiet=False)

if not os.path.exists(zip_path) or not zipfile.is_zipfile(zip_path):
    raise RuntimeError(
        "Download failed. Check that Google Drive access is set to Anyone with the link."
    )

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_path)

print("Dataset extracted successfully.")
for root, dirs, files in os.walk(extract_path):
    if "data.yaml" in files:
        print("Dataset configuration:", os.path.join(root, "data.yaml"))
```
